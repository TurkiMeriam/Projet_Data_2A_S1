```python
import pandas as pd

df = pd.read_parquet("/home/onyxia/work/Projet_Data_2A_S1/data/processed/dvf_appartements_stations.parquet")

print("Nb lignes :", len(df))
print("Valeurs manquantes sur dist_station_m :", df["dist_station_m"].isna().sum())
print("\nRépartition par classe de distance :")
print(df["classe_dist"].value_counts(dropna=False))
print("\nPrix moyen au m² selon la classe de distance :")
print(df.groupby("classe_dist", observed=False)["prix_m2"].agg(["count", "mean", "median"]))
```

```python
import pandas as pd
from pathlib import Path

dossier = Path("/home/onyxia/work/Projet_Data_2A_S1/data/processed/")     

fichiers = sorted(dossier.glob("dvf_appartements_20??.csv"))
print("Fichiers trouvés :", [f.name for f in fichiers])

apps = pd.concat(
    [pd.read_csv(f, dtype={"code_departement": str, "code_commune": str}, parse_dates=["date_mutation"])
     for f in fichiers],
    ignore_index=True,
)
apps["zone"] = apps["code_departement"].str.zfill(2).map(lambda d: "Paris" if d == "75" else "Petite couronne")

print(apps.shape)
print(apps.groupby(["zone", "annee"]).size().unstack())
print(apps.groupby("code_departement").size())
print(apps[["prix_m2", "surface_reelle_bati"]].describe().round(1))
```

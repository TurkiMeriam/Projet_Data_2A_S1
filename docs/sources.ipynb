## * Inventaire des Sources de Données

---

### 1. Valeurs Foncières (Immobilier)
* **DVF géolocalisées (Etalab)** — [data.gouv.fr – DVF géolocalisées](https://www.data.gouv.fr/datasets/demandes-de-valeurs-foncieres-geolocalisees/discussions)
  * **Remarque :** Un fichier `.csv` par année et par département. Sélectionner uniquement les départements de la petite couronne : **75, 92, 93 et 94**.
* **Variante DVF+ (Cerema)** *(Optionnel)* — [data.gouv.fr – DVF+ open-data](https://www.data.gouv.fr/datasets/dvf-open-data)

---

### 2. Réseau Ferré & Géographie des Transports (IDFM)
* **Gares et stations IDFM** — [Gares et stations du réseau ferré d'IDF (par ligne)](https://www.data.gouv.fr/datasets/gares-et-stations-du-reseau-ferre-dile-de-france-par-ligne)
  * **Remarque :** Contient un point par station *et* par ligne. 
---

### 3. Fréquentation des Transports
* **Option IDFM** — [Validations sur le réseau ferré, par jour (T4 2025)](https://www.data.gouv.fr/datasets/validations-sur-le-reseau-ferre-nombre-de-validations-par-jour-4eme-trimestre-2025)
  * **Remarque :** Plus récent et exhaustif (**couvre tous les opérateurs : RATP et SNCF**). Disponible sous forme d'un fichier par trimestre, également accessible sur [data.iledefrance.fr](https://data.iledefrance.fr/explore/dataset/validations-sur-le-reseau-ferre-nombre-de-validations-par-jour-1er-trimestre-2025/).
* **Option RATP** — [Trafic annuel entrant par station 2019](https://data.gouv.fr/en/datasets/trafic-annuel-entrant-par-station-du-reseau-ferre-2019)
  * **Remarque :** Couvre le **métro et le RER RATP uniquement**. D'autres millésimes sont disponibles : [2018](https://www.data.gouv.fr/datasets/trafic-annuel-entrant-par-station-du-reseau-ferre-2018), [2016](https://data.gouv.fr/en/datasets/trafic-annuel-entrant-par-station-du-reseau-ferre-2016-1)…

---

### 4. Découpage Territorial, Aménités & Environnement
* **Contours IRIS (IGN / INSEE)** — [data.gouv.fr – Contours IRIS INSEE-IGN](https://www.data.gouv.fr/datasets/contours-iris-insee-ign)
* **BPE — Base Permanente des Équipements (INSEE)** — [Équipements géolocalisés 2025 (BPE)](https://www.insee.fr/fr/statistiques/8217525)
  * **Remarque :** Version géolocalisée, documentation complète sur la [page BPE 2024](https://www.insee.fr/fr/metadonnees/source/operation/s2216/bases-donnees-ligne).
* **OpenStreetMap (OSM)** — [Extrait Île-de-France (Geofabrik)](https://download.geofabrik.de/europe/france/ile-de-france.html)
